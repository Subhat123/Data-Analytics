Find where the mark is 78

Search

Conditional-based searching

```python
np.where(marks==78)
```

```python
mark=np.array([[85,78,90],
      [92,85,98],
      [76,82,80]])

print(np.where(mark==78))
print(mark)
```

Positions

```python
mark=np.array([[85,78,90],
      [92,85,98],
      [76,82,80]])

np.where(mark>=80) # gives me indices
print(mark)
```

```python
mark=np.array([[85,78,90],
      [92,85,98],
      [76,82,80]])

print(mark[np.where(mark>=80)]) # ---> gives exact values
print(mark)
```
