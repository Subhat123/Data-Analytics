```python
cond=df[df["Attendance"] >= 75]
cond.Name
```

Find students whose attendance is atleast 75 AND Python mark is atleast 80

# Manipulation

Calculate the average of Python,SQL and Excel

```python
df["Python"]+df["SQL"]+df["Excel"]//3
```

```python
(df["Python"]+df["SQL"]+df["Excel"])//3
```

```python
df["Average"]=((df["Python"]+df["SQL"]+df["Excel"])//3)
df
```

```python
df["Results"]=df["Average"].apply( lambda avg:
   "Pass" if avg>=50 else "Fail"
)
df
```

# GroupBy

What is the average Python mark for each department?

```python
df.groupby("Department")["Python"].mean()
```

Number of students per department

```python
df.groupby ("Department").size()
```

Highest Python mark by department

```python
df.groupby("Department")["Python"].max()
```

# GroupBy with Multiple metrices

```python
df.groupby("Department").agg(
    {
        "Student_ID":"count",
        "Python":"mean",
        "Excel":"mean",
        "SQL":"mean"
    }
)
```

```python
df.groupby("Department").agg(
    {
        "Student_ID":"count",
        "Python":"mean",
        "Excel":"mean",
        "SQL":"mean",
        "Attendance":"mean"
    }
)
```
